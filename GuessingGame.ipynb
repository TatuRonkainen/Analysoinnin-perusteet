```python
answer = "Tatu"
count = 0 

while True:
    guess = input("Guess my name:")
    count += 1

    if guess == answer:
        print("Congratulations")
        print("You guessed:",count, "times")
        break

    else:
        
        quit = input("Do you want to quit? (y/n) ")
        if quit == "y":
            break
        print("Guess again")







```
