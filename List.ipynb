```python
import random

thrownDiceNumbers = []

for i in range(5):
    thrownDiceNumbers.append(random.randint(1,6))

print(thrownDiceNumbers)


print(sum(thrownDiceNumbers))

print(max(thrownDiceNumbers))
```
